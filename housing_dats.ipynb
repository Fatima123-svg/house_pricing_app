```python
X = df[["No of bedroms"]]
y = df["Price"]
```

```python
plt.scatter(X,y)
```

```python
predicted = model.predict(X)
```

```python
predicted = model.predict(X)

plt.scatter(X,y)
plt.plot(X,predicted,color="red")
plt.xlabel("Area")
plt.ylabel("Sales")
plt.title("Area vs Saales")
plt.show()
```
